# Problema 1 — Classificação da média nas fronteiras

**Nível:** Moderado


Um sistema acadêmico recebe três notas de um aluno.

Crie um programa que:

1. leia as três notas como `float`;
2. calcule a média aritmética;
3. classifique:
   - média menor que `4`: `Reprovado`;
   - média maior ou igual a `4` e menor que `6`: `Recuperação`;
   - média maior ou igual a `6` e menor que `9`: `Aprovado`;
   - média maior ou igual a `9`: `Aprovado com destaque`;
4. exiba a média com duas casas decimais e a classificação.

### Casos obrigatórios de teste

Sua lógica deve classificar corretamente médias exatamente iguais a:

- `4.0`
- `6.0`
- `9.0`

### Microdefesa

Explique em uma linha por que a ordem dos testes pode alterar o resultado.

```python
# ============================================================
# PROBLEMA 1
# ============================================================

# PREVISÃO E DECOMPOSIÇÃO
# Entrada(s):
# Três notas do aluno.
#
# Processamento:
# Calcular a média das três notas e verificar em qual faixa ela está.
#
# Saída:
# A média com duas casas decimais e a classificação.

# IMPLEMENTAÇÃO

print("Olá, para calcularmos a sua nota preencha as informações abaixo!")
print()

nome = str(input("Digite o seu nome: "))
print()

nota1 = float(input("Digite a primeira nota: "))
nota2 = float(input("Digite a segunda nota: "))
nota3 = float(input("Digite a terceira nota: "))

media = (nota1 + nota2 + nota3) / 3

if media < 4:
    classificacao = "Reprovado"
elif media < 6:
    classificacao = "Recuperação"
elif media < 9:
    classificacao = "Aprovado"
else:
    classificacao = "Aprovado com destaque"

print(f"Sua média: {media:.2f}")
print(f"Classificação: {classificacao}")
print(f"Obrigado por usar nosso sistema {nome}!")


# MICRODEFESA: por que a ordem das condições pode alterar a classificação?
# Porque o programa executa a primeira condição verdadeira, então uma condição
# mais ampla colocada antes pode impedir que uma classificação correta seja alcançada.
```

O programa deve:

- informar `Todos iguais` se os três valores forem iguais;
- caso contrário, exibir o maior valor;
- exibir o menor valor.

### Atenção: empates parciais são válidos

Seu programa deve funcionar também para:

- `8, 8, 3`
- `4, 9, 9`
- `5, 2, 5`

### Restrição

Não utilize `min()` nem `max()`.

### Microdefesa

Explique por que testar apenas `a > b and a > c` não é suficiente para todos os casos.

```python
# ============================================================
# PROBLEMA 3
# ============================================================

# PREVISÃO E DECOMPOSIÇÃO
# Entrada(s):
# Três números inteiros: a, b e c.
#
# Processamento:
# Verificar se todos são iguais. Caso contrário, comparar os valores
# para encontrar o maior e o menor.
#
# Saída:
# Todos iguais ou o maior e o menor valor.

# IMPLEMENTAÇÃO

a = int(input("Digite o valor de a: "))
b = int(input("Digite o valor de b: "))
c = int(input("Digite o valor de c: "))

if a == b and b == c:
    print("Todos iguais")
else:
    if a >= b and a >= c:
        maior = a
    elif b >= a and b >= c:
        maior = b
    else:
        maior = c

    if a <= b and a <= c:
        menor = a
    elif b <= a and b <= c:
        menor = b
    else:
        menor = c

    print(f"Maior valor: {maior}")
    print(f"Menor valor: {menor}")


# MICRODEFESA: o que muda quando dois valores máximos ou mínimos são iguais?
# Devemos usar >= e <= para que valores empatados também possam ser identificados
# corretamente como maior ou menor.
```

Identifique qual atribuição faz o laço se aproximar da condição de término.

```python
# ============================================================
# PROBLEMA 5
# ============================================================

# PREVISÃO E DECOMPOSIÇÃO
# Entrada(s):
# Um número inteiro n maior ou igual a 0.
#
# Processamento:
# Fazer uma contagem regressiva usando while e verificar os múltiplos de 5.
#
# Saída:
# Os números de n até 0, indicando os positivos divisíveis por 5.

# IMPLEMENTAÇÃO

n = int(input("Digite um número maior ou igual a 0: "))

while n >= 0:
    if n > 0 and n % 5 == 0:
        print(f"{n} é divisível por 5")
    else:
        print(n)

    n = n - 1


# MICRODEFESA: qual linha garante que o while possa terminar?
# A linha n = n - 1 diminui o valor de n a cada repetição, aproximando-o
# da condição de término.
```

**Nível:** Moderado


Receba:

- `limite`, inteiro maior ou igual a `1`;
- `divisor`, inteiro diferente de `0`.

Some todos os inteiros de `1` até `limite`, inclusive, que sejam divisíveis por `divisor`.

### Exemplo

Para `limite = 10` e `divisor = 3`:

`3 + 6 + 9 = 18`

### Restrições

- utilize `for` e `range`;
- não utilize `sum()`.

### Casos obrigatórios de teste

- `limite = 10`, `divisor = 3`
- `limite = 5`, `divisor = 7`
- `limite = 6`, `divisor = 2`

Explique por que o acumulador deve possuir um valor inicial antes do `for`.

```python
# ============================================================
# PROBLEMA 6
# ============================================================

# PREVISÃO E DECOMPOSIÇÃO
# Entrada(s):
# Um limite e um divisor.
#
# Processamento:
# Percorrer os números de 1 até o limite e somar apenas os divisíveis pelo divisor.
#
# Saída:
# A soma dos números divisíveis pelo divisor.

# IMPLEMENTAÇÃO

limite = int(input("Digite o limite: "))
divisor = int(input("Digite o divisor: "))

soma = 0

for numero in range(1, limite + 1):
    if numero % divisor == 0:
        soma = soma + numero

print(f"Soma: {soma}")


# MICRODEFESA: por que o acumulador precisa ser inicializado?
# Porque ele precisa começar com um valor conhecido, neste caso 0, antes de receber
# os valores que serão somados durante o for.
```

# Problema 10 — Função de classificação numérica

**Nível:** Moderado


Crie a função:

`classificar_numero(numero)`

Ela deve retornar exatamente uma das strings:

- `positivo e par`
- `positivo e ímpar`
- `negativo e par`
- `negativo e ímpar`
- `zero`

Depois:

1. leia um inteiro;
2. passe-o como argumento;
3. armazene o retorno;
4. exiba o resultado.

### Casos obrigatórios de teste

- `-4`
- `-3`
- `0`
- `2`
- `5`

### Microdefesa

Explique por que `0` merece uma decisão própria antes de classificá-lo como positivo ou negativo.

```python
# ============================================================
# PROBLEMA 10
# ============================================================

# PREVISÃO E DECOMPOSIÇÃO
# Entrada(s):
# Um número inteiro.
#
# Processamento:
# Verificar se o número é zero.
# Caso não seja, verificar se é positivo ou negativo
# e depois verificar se é par ou ímpar.
#
# Saída:
# Uma classificação do número:
# "positivo e par", "positivo e ímpar",
# "negativo e par", "negativo e ímpar" ou "zero".
#

# IMPLEMENTAÇÃO

def classificar_numero(numero):
    if numero == 0:
        return "zero"
    elif numero > 0 and numero % 2 == 0:
        return "positivo e par"
    elif numero > 0 and numero % 2 != 0:
        return "positivo e ímpar"
    elif numero < 0 and numero % 2 == 0:
        return "negativo e par"
    else:
        return "negativo e ímpar"


numero = int(input("Digite um número inteiro: "))

resultado = classificar_numero(numero)

print(resultado)


# MICRODEFESA: por que zero não deve cair nas categorias positivo/negativo?
# Porque o zero não é positivo nem negativo. Ele é um número neutro.
# Por isso, ele precisa ser verificado primeiro e retornar "zero".
```

# Problema 11 — Calculadora e retorno None

**Nível:** Moderado/Difícil


Crie:

`calcular(a, b, operacao)`

`operacao` será uma `str` contendo exatamente um destes símbolos:

- `+`
- `-`
- `*`
- `/`

A função deve:

- retornar o resultado da operação válida;
- retornar `None` se a operação não for reconhecida;
- retornar `None` se houver tentativa de divisão por zero.

Depois, o programa deve exibir:

`Operação inválida`

quando o retorno for `None`.

### Casos obrigatórios de teste

- `10, 2, "+"`
- `10, 2, "/"`
- `10, 0, "/"`
- `10, 2, "x"`

### Microdefesa

Explique a diferença entre imprimir o resultado dentro da função e retorná-lo.

```python
# ============================================================
# PROBLEMA 11
# ============================================================

# PREVISÃO E DECOMPOSIÇÃO
# Entrada(s):
# Dois números e uma operação (+, -, *, /).
#
# Processamento:
# Verificar qual operação foi escolhida.
# Realizar a operação correspondente.
# Caso a operação seja inválida ou haja divisão por zero,
# retornar None.
#
# Saída:
# O resultado da operação ou "Operação inválida".
#

# IMPLEMENTAÇÃO

def calcular(a, b, operacao):
    if operacao == "+":
        return a + b
    elif operacao == "-":
        return a - b
    elif operacao == "*":
        return a * b
    elif operacao == "/":
        if b == 0:
            return None
        return a / b
    else:
        return None


a = float(input("Digite o primeiro número: "))
b = float(input("Digite o segundo número: "))
operacao = input("Digite a operação (+, -, *, /): ")

resultado = calcular(a, b, operacao)

if resultado is None:
    print("Operação inválida")
else:
    print(resultado)


# MICRODEFESA: Explique a diferença entre imprimir o resultado
# dentro da função e retorná-lo.
# Imprimir mostra o resultado na tela, mas não permite que o
# programa utilize esse valor posteriormente.
# Retornar com return entrega o resultado para quem chamou a
# função, permitindo armazená-lo em uma variável e fazer outras
# operações com ele.
```

# Problema 12 — Potência por multiplicações sucessivas

**Nível:** Difícil


Crie:

`potencia(base, expoente)`

A função deve calcular a potência usando multiplicações sucessivas.

Considere `expoente` inteiro maior ou igual a `0`.

### Restrições

- não utilize `**`;
- não utilize bibliotecas.

### Casos obrigatórios de teste

- `potencia(2, 4)` → `16`
- `potencia(5, 0)` → `1`
- `potencia(-2, 3)` → `-8`
- `potencia(-2, 4)` → `16`

### Microdefesa

Explique por que o resultado inicial precisa ser `1`, especialmente quando o expoente é `0`.

```python
# ============================================================
# PROBLEMA 12
# ============================================================

# PREVISÃO E DECOMPOSIÇÃO
# Entrada(s):
# Uma base e um expoente inteiro maior ou igual a 0.
#
# Processamento:
# Começar o resultado com 1.
# Multiplicar o resultado pela base a cada repetição,
# até completar a quantidade indicada pelo expoente.
#
# Saída:
# O resultado da potência.
#

# IMPLEMENTAÇÃO

def potencia(base, expoente):
    resultado = 1

    for i in range(expoente):
        resultado = resultado * base

    return resultado


print(potencia(2, 4))
print(potencia(5, 0))
print(potencia(-2, 3))
print(potencia(-2, 4))


# MICRODEFESA: por que o resultado inicial precisa ser 1,
# especialmente quando o expoente é 0?
# Porque qualquer número elevado a 0 é igual a 1.
# Além disso, 1 é o valor neutro da multiplicação, então
# ele permite que a primeira multiplicação seja feita
# corretamente. Quando o expoente é 0, o laço não executa
# nenhuma multiplicação e o resultado continua sendo 1.
```

# Problema 13 — Soma dos dígitos sem str

**Nível:** Difícil


Receba um inteiro positivo e calcule a soma de seus dígitos **sem converter o número para `str`**.

### Exemplo

`5832` → `5 + 8 + 3 + 2` → `18`

### Restrições

- não utilize `str`;
- não utilize coleções;
- não utilize métodos de string;
- utilize operações aritméticas e repetição.

### Casos obrigatórios de teste

- `10` → `1`
- `999` → `27`
- `1001` → `2`
- `5832` → `18`

### Microdefesa

Explique o papel de `% 10` e de `// 10` no algoritmo.

```python
# ============================================================
# PROBLEMA 13
# ============================================================

# PREVISÃO E DECOMPOSIÇÃO
# Entrada(s):
# Um número inteiro positivo.
#
# Processamento:
# Extrair o último dígito usando % 10, somá-lo ao acumulador
# e remover esse dígito usando // 10.
#
# Saída:
# A soma dos dígitos.

# IMPLEMENTAÇÃO

numero = int(input("Digite um número inteiro positivo: "))

soma = 0

while numero > 0:
    digito = numero % 10
    soma = soma + digito
    numero = numero // 10

print(f"Soma dos dígitos: {soma}")


# MICRODEFESA: o que % 10 extrai e o que // 10 remove?
# % 10 extrai o último dígito do número e // 10 remove esse último dígito.
```

### Microdefesa

Explique por que modificar diretamente o único valor de `numero` durante a inversão pode impedir a comparação final correta.

```python
# ============================================================
# PROBLEMA 15
# ============================================================

# PREVISÃO E DECOMPOSIÇÃO
# Entrada(s):
# Um número inteiro positivo.
#
# Processamento:
# Preservar o número original.
# Criar uma cópia para realizar a inversão.
# Retirar os últimos dígitos usando % 10.
# Construir o número invertido usando multiplicações por 10.
# Comparar o número invertido com o original.
#
# Saída:
# True se o número for palíndromo.
# False se não for palíndromo.
#

# IMPLEMENTAÇÃO

def eh_palindromo(numero):
    original = numero
    invertido = 0

    while numero > 0:
        digito = numero % 10
        invertido = invertido * 10 + digito
        numero = numero // 10

    return invertido == original


numero = int(input("Digite um número positivo: "))

if eh_palindromo(numero):
    print("Palíndromo")
else:
    print("Não palíndromo")


# MICRODEFESA: por que modificar diretamente o único valor de
# numero durante a inversão pode impedir a comparação final correta?
# Porque, durante a inversão, o valor de numero vai sendo reduzido
# até chegar a 0. Se não guardarmos o valor original em outra
# variável, não teremos mais o número inicial para comparar com
# o valor invertido. Por isso, usamos a variável "original".
```
