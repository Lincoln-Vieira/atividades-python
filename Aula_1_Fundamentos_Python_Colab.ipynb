# 1. Lógica de programação

Programar começa antes do código.

Considere a regra:

> Um aluno é aprovado quando sua média é maior ou igual a 6.

A lógica é:

1. obter a média;
2. comparar a média com 6;
3. decidir o resultado;
4. apresentar a resposta.

````python
media = 7.5

if media >= 6:
    print("Aprovado")
else:
    print("Não aprovado")
````

# 2. Decomposição de problemas

Um problema maior pode ser dividido em partes menores.

Problema:

> Receber três notas, calcular a média e informar a situação.

Decomposição:

1. receber as notas;
2. calcular a média;
3. comparar com a nota mínima;
4. definir a situação;
5. exibir o resultado.

````python
nota1 = 8.0
nota2 = 7.0
nota3 = 9.0

media = (nota1 + nota2 + nota3) / 3

if media >= 6:
    situacao = "Aprovado"
else:
    situacao = "Não aprovado"

print(f"Média: {media:.2f}")
print(f"Situação: {situacao}")
````

### Pausa prática

Altere apenas uma das notas até fazer o resultado mudar de `Aprovado` para `Não aprovado`.

Depois responda mentalmente: **qual parte do problema mudou — entrada, processamento ou decisão?**

# 3. Algoritmos

Um algoritmo descreve os passos necessários para executar uma tarefa.

Exemplo:

> Calcular 10% de desconto.

Passos:

1. obter o valor;

3. subtrair o desconto;
4. exibir o resultado.

````python
valor = 200
desconto = valor * 0.10
valor_final = valor - desconto

print(valor_final)
````

Uma variável definida dentro de uma função não deve ser considerada automaticamente disponível fora dela.
